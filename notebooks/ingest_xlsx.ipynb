```python
from pathlib import Path
import pandas as pd
import secrets

from src.objects import Singer

page_defs: dict[str, str] = {"A1": "a1", "A2": "a2", "BARI": "b", "SOP 1": "s1", "SOP 2": "s2"}

singers: list[Singer] = []

for page_name, singer_type in page_defs.items():
    data = pd.read_excel(Path("./data/staging.xlsx"), sheet_name=page_name, header=None).fillna("")

    for r in data.iterrows():
        uid = secrets.token_hex(4)
        singers.append(Singer(uid, str(r[1][0]).title(), singer_type, notes=r[1][2]))

from pprint import pprint

pprint(singers, indent = 2)
```

```python
import json
from dataclasses import asdict

with open(Path("./data/singers.json"), "w") as f:
    f.write(json.dumps([asdict(s) for s in singers], indent=2))
```
